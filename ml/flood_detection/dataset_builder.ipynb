# SAR Dataset Builder Notebook

This notebook demonstrates how to read a Sentinel-1 GRD SAR file, visualize its VV and VH bands, and create placeholder training pairs for flood detection. You will need a sample Sentinel-1 GRD GeoTIFF file (e.g., from ESA Copernicus Open Access Hub or ASF DAAC).

## Required Libraries

Make sure you have the following Python libraries installed:
- `rasterio`: For reading and writing geospatial raster datasets.
- `numpy`: For numerical operations.
- `matplotlib`: For visualization.

```bash
pip install rasterio numpy matplotlib
```

## Sample Sentinel-1 GRD Data

Sentinel-1 Level-1 GRD (Ground Range Detected) data can be downloaded from various sources. You will typically need an account for these platforms:
- [ESA Copernicus Open Access Hub](https://scihub.copernicus.eu/) (Requires registration)
- [Alaska Satellite Facility (ASF) DAAC](https://search.asf.alaska.edu/) (Requires registration)

**Note:** This notebook assumes the input SAR file is a GeoTIFF with VV and VH bands as band 1 and band 2 respectively. Please ensure your sample data matches this format.


````python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
import os

# --- Configuration ---
sample_sar_file = "path/to/your/sentinel1_grd_sample.tif"  # IMPORTANT: Update this path to your actual SAR file
output_dataset_dir = "../dataset/"

# Create output directory if it doesn't exist
os.makedirs(output_dataset_dir, exist_ok=True)

print(f"Using sample SAR file: {sample_sar_file}")
print(f"Output dataset directory: {output_dataset_dir}")

if not os.path.exists(sample_sar_file):
    print(f"ERROR: Sample SAR file not found at {sample_sar_file}. Please update `sample_sar_file` variable with a valid path.")
else:
    print("Sample SAR file found. Proceeding with processing.")

````

## 1. Read and Inspect SAR Image

We'll use `rasterio` to open the SAR GeoTIFF file and inspect its metadata, such as coordinate reference system (CRS), transform, and band count.


````python
if os.path.exists(sample_sar_file):
    with rasterio.open(sample_sar_file) as src:
        print(f"Image width: {src.width}, height: {src.height}")
        print(f"Number of bands: {src.count}")
        print(f"CRS: {src.crs}")
        print(f"Transform: {src.transform}")
        print(f"Band 1 (VV) description: {src.descriptions[0] if src.descriptions else 'N/A'}")
        print(f"Band 2 (VH) description: {src.descriptions[1] if len(src.descriptions) > 1 else 'N/A'}")

        # Read bands
        vv_band = src.read(1)
        vh_band = src.read(2) if src.count > 1 else None

        print("VV band statistics:")
        print(f"  Min: {np.min(vv_band)}, Max: {np.max(vv_band)}, Mean: {np.mean(vv_band)}")

        if vh_band is not None:
            print("VH band statistics:")
            print(f"  Min: {np.min(vh_band)}, Max: {np.max(vh_band)}, Mean: {np.mean(vh_band)}")
else:
    print("Skipping SAR image inspection as the sample file was not found.")
````

## 2. Visualize VV and VH Bands

We'll visualize the VV and VH bands to understand the raw data. VV (Vertical transmit/Vertical receive) and VH (Vertical transmit/Horizontal receive) polarizations provide different information about the surface, which is crucial for flood detection.







````python
if os.path.exists(sample_sar_file):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

    ax1.imshow(vv_band, cmap='gray')
    ax1.set_title('VV Band')
    ax1.axis('off')

    if vh_band is not None:
        ax2.imshow(vh_band, cmap='gray')
        ax2.set_title('VH Band')
        ax2.axis('off')
    else:
        ax2.set_title('VH Band Not Available')
        ax2.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Skipping band visualization as the sample file was not found.")
````

## 3. Create Training Pairs (Image, Mask) Placeholder

This section outlines the process of creating training pairs. For flood detection, each pair typically consists of a preprocessed SAR image and a corresponding flood mask. In this placeholder, we will simulate this process. You would replace the mask creation with actual flood mask generation based on your ground truth data.


````python
if os.path.exists(sample_sar_file):
    # For demonstration, we'll use a simple thresholding for a placeholder mask
    # In a real scenario, this would come from labeled data or an external source.
    print("Creating placeholder flood mask...")
    flood_mask = (vv_band < np.mean(vv_band) * 0.5).astype(np.uint8) # Simple thresholding example

    # Normalize the SAR image bands (simple min-max for visualization)
    normalized_vv = (vv_band - np.min(vv_band)) / (np.max(vv_band) - np.min(vv_band))
    normalized_vh = (vh_band - np.min(vh_band)) / (np.max(vh_band) - np.min(vh_band)) if vh_band is not None else np.zeros_like(normalized_vv)

    # Stack normalized bands into an image array
    sar_image_processed = np.stack([normalized_vv, normalized_vh], axis=-1)

    # Visualize the image and mask
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
    ax1.imshow(sar_image_processed[:,:,0], cmap='gray') # Show VV band as example
    ax1.set_title('Processed SAR Image (VV band)')
    ax1.axis('off')

    ax2.imshow(flood_mask, cmap='viridis', alpha=0.7) # Visualize mask
    ax2.set_title('Placeholder Flood Mask')
    ax2.axis('off')

    plt.tight_layout()
    plt.show()

    # Save example training pair (placeholder for more sophisticated saving)
    image_output_path = os.path.join(output_dataset_dir, "sample_image.npy")
    mask_output_path = os.path.join(output_dataset_dir, "sample_mask.npy")

    np.save(image_output_path, sar_image_processed)
    np.save(mask_output_path, flood_mask)

    print(f"Saved sample image to: {image_output_path}")
    print(f"Saved sample mask to: {mask_output_path}")
else:
    print("Skipping training pair creation as the sample file was not found.")
````
